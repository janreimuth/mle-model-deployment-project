```python
import sys
sys.path.append("../src")

import pandas as pd

from features import (
    load_feature_columns,
    prepare_features,
    prepare_target,
)

ROOT = "../data/"

train = pd.read_csv(ROOT + "early_warning_train.csv")
validation = pd.read_csv(ROOT + "early_warning_validation.csv")

feature_columns = load_feature_columns(
    ROOT + "feature_columns.txt"
)

X_train = prepare_features(train, feature_columns)
y_train = prepare_target(train)

X_val = prepare_features(validation, feature_columns)
y_val = prepare_target(validation)

print("X_train:", X_train.shape)
print("y_train:", y_train.shape)
print("X_val:", X_val.shape)
print("y_val:", y_val.shape)
```

```python
from sklearn.ensemble import RandomForestClassifier

model = RandomForestClassifier(
    n_estimators=200,
    random_state=42,
    n_jobs=-1,
)

model.fit(X_train, y_train)
```

```python
y_pred_proba = model.predict_proba(X_val)[:, 1]

print(y_pred_proba[:10])
```

```python
from sklearn.metrics import root_mean_squared_error

rmse = root_mean_squared_error(
    y_val,
    y_pred_proba,
)

print(f"Validation RMSE: {rmse:.4f}")
```

```python
print("Actual positive rate:", y_val.mean())
print("Mean predicted probability:", y_pred_proba.mean())

print(
    pd.Series(y_pred_proba).describe()
)

comparison = pd.DataFrame({
    "actual": y_val.values,
    "predicted_probability": y_pred_proba,
})

comparison.head(20)
```

```python
import mlflow
import mlflow.sklearn

with mlflow.start_run():

    model = RandomForestClassifier(
        n_estimators=200,
        random_state=42,
        n_jobs=-1,
    )

    model.fit(X_train, y_train)

    y_pred_proba = model.predict_proba(X_val)[:, 1]

    rmse = root_mean_squared_error(
        y_val,
        y_pred_proba,
    )

    mlflow.log_param(
        "model_type",
        "RandomForestClassifier",
    )
    mlflow.log_param(
        "n_estimators",
        200,
    )
    mlflow.log_param(
        "random_state",
        42,
    )
    mlflow.log_param(
        "n_features",
        len(feature_columns),
    )

    mlflow.log_metric(
        "validation_rmse",
        rmse,
    )

    mlflow.sklearn.log_model(
        model,
        name="model",
        skops_trusted_types=[
            "sklearn.tree._tree.Tree"
        ],
    )

    print(f"Validation RMSE: {rmse:.4f}")
```
